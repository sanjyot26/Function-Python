Q.6 Explain the concept of generators in Python and how they are defined.

Ans > Generator are the function which use `yield` instead of `return` they produce value one at a time and useful for large dataset

```python
def gen_numbers():
    for i in range(3):
        yield i

for x in gen_numbers():
    print(x)
```

Q.10  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Ans > `map()` >> map compution to every element in iterable .

 `reduce()` >> reduce the element to a singe value output.

 `filter()` >> filter out element based on given condition .

```python
from functools import reduce

nums = [1, 2, 3, 4]
print(list(map(lambda x: x*2, nums)))          # [2, 4, 6, 8]
print(list(filter(lambda x: x%2==0, nums)))    # [2, 4]
print(reduce(lambda x,y: x+y, nums))           # 10
```

Q.1  Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

```python
def sum_even(nums):
    return sum([n for n in nums if n % 2 == 0])

print(sum_even([1,2,3,4,5,6]))
```

Q.4 Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [x for x in range(1,201) if is_prime(x)]
print(primes)
```

Q.5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.


```python
def fib(n):
    a = 0
    b = 1
    for i in range(n): #if n=5, 0, 1, 2, 3, 4
        yield a
        a, b = b, a+b
```

Q.6 . Write a generator function in Python that yields the powers of 2 up to a given exponent.


```python
def powers_of_two(n):
    for i in range(n+1):
        yield 2**i

print(list(powers_of_two(5)))
```

Q.7  Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_file(filename):
    with open(filename) as f:
        for line in f:
            yield line.strip()
```

Q.8  Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

```python
data = [(1,3), (2,1), (3,2)]
data.sort(key=lambda x: x[1])
print(data)
```

Q.9 Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
celsius = [0, 20, 30, 40]
fahrenheit = list(map(lambda c: (c*9/5)+32, celsius))
print(fahrenheit)
```

Q.10  Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
s = "Sanjyot Sawant"
result = "".join(filter(lambda x: x.lower() not in "aeiou", s))
print(result)
```
